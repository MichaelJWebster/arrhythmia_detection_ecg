# ECG classification dataset

The data for this notebook can be obtained from [Kaggle ECG hearbeat classification dataset](https://www.kaggle.com/datasets/shayanfazeli/heartbeat). Place the csv files in a data_directory and change the location of DATA_DIR in the next cell to reference it.


```python
from pathlib import Path
import os
import random

import torch
import numpy as np
import pandas as pd
import albumentations as A
import matplotlib.pyplot as plt
from torch.utils.data import Dataset, DataLoader, Subset, random_split
from torch.optim import lr_scheduler
from typing import cast

import pytorch_lightning as pl

from ecg_data import ECGDataset, read_ecg_csv, default_num_workers, BeatAugment, balanced_sampler
from ecg_models import EcgLightningModule, make_ecg_model

# Architecture and training recipe (see ecg_models.py); set ECG_ARCH to choose.
#   "paper":  Kachuee et al. (2018), Fig. 2: residual CNN; Adam, learning rate x0.75 every
#             10,000 steps; rare classes oversampled to balance, with random augmentation.
#   "simple": the earlier 4-convolution CNN; AdamW + ReduceLROnPlateau; class-weighted loss.
ARCH = os.environ.get("ECG_ARCH", "paper")
# How class imbalance is handled; set ECG_BALANCE to override the architecture's default.
#   "augment": balanced sampling + BeatAugment, unweighted loss.  "weights": class-weighted loss.
BALANCE = os.environ.get("ECG_BALANCE", "augment" if ARCH == "paper" else "weights")
print(f"ARCH={ARCH}  BALANCE={BALANCE}")

# Data directory holding mitbih_*.csv and ptbdb_*.csv. Defaults to ./data;
# set ECG_DATA_DIR to point elsewhere.
DATA_DIR = os.environ.get("ECG_DATA_DIR", "./data")
```

Set random number seed:

```python
SEED = 42

os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
torch.use_deterministic_algorithms(True, warn_only=True)

# Use Tensor Cores on Ampere+ GPUs (e.g. RTX 3000/4000 series) for faster matmuls
torch.set_float32_matmul_precision("high")

pl.seed_everything(SEED, workers=True)

os.environ['CUBLAS_WORKSPACE_CONFIG'] = ':4096:8'
torch.use_deterministic_algorithms(True)
```

Create the main train dataset, and break down into train and validation subsets using random_split.

Make sure to set a random seed before doing this so we're repeatable.

```python
# Split row indices (seeded, so the split is repeatable), then build the datasets, so the
# training split can be augmented while validation, test and evaluation copies are not.
train_split, val_split = random_split(range(len(td)), [0.8, 0.2], generator=torch.Generator().manual_seed(SEED))
train_idx, val_idx = list(train_split), list(val_split)
train_labels = td.iloc[train_idx, -1].to_numpy().astype(int)

if BALANCE == "augment":
    train_ds = ECGDataset(td.iloc[train_idx], transform=BeatAugment())
    train_sampler = balanced_sampler(train_labels, seed=SEED)
else:
    train_ds = ECGDataset(td.iloc[train_idx])
    train_sampler = None
val_ds = ECGDataset(td.iloc[val_idx])
train_eval_ds = ECGDataset(td.iloc[train_idx])  # un-augmented, for scoring the training split

NUM_WORKERS = default_num_workers()
loader_kwargs = dict(batch_size=64, num_workers=NUM_WORKERS, persistent_workers=NUM_WORKERS > 0)
train_dl = DataLoader(train_ds, sampler=train_sampler, shuffle=train_sampler is None, **loader_kwargs)
val_dl = DataLoader(val_ds, shuffle=False, **loader_kwargs)
test_dl = DataLoader(ECGDataset(test_data), shuffle=False, **loader_kwargs)
```

## Build the model

Both architectures are defined in `ecg_models.py`, shared with `transfer_learning.ipynb`:

- **`paper`**: the network in Fig. 2 of Kachuee et al. (2018). A convolution, then five residual blocks (two convolutions of 32 kernels of size 5, a skip connection, ReLU and max-pooling of size 5 / stride 2), then two fully connected layers: 13 weight layers, about 54k parameters.
- **`simple`**: the earlier model, adapted from [this GitHub implementation](https://github.com/sajiddeboss/ECG-HeartBeat-Classification/blob/main/Model.ipynb): four plain convolutions and a large dense head, about 668k parameters.

```python
preview = make_ecg_model(ARCH)
print(preview)
print(f"{ARCH}: {sum(p.numel() for p in preview.parameters()):,} parameters")
```

`EcgLightningModule` (in `ecg_models.py`) wraps the model for training; its optimiser and learning-rate schedule follow the architecture.

Class imbalance (most beats are N) is handled one of two ways:

- **`augment`** (the paper's approach): training beats are drawn with every class equally likely, rare beats are repeated with random augmentation (`BeatAugment`), and the loss is unweighted. The paper does not say how it augmented; amplitude scaling, time stretching and light noise are my choices.
- **`weights`**: natural sampling, with inverse-frequency class weights in the loss.

```python
class_counts = np.bincount(train_labels, minlength=5)
print("Class counts:", dict(zip(ECGDataset.VAL_TO_CLASS.values(), class_counts)))

if BALANCE == "weights":
    # Inverse-frequency class weights, normalised so they sum to the number of classes
    class_weights = 1.0 / class_counts
    class_weights = class_weights / class_weights.sum() * len(class_weights)
    print("Class weights:", dict(zip(ECGDataset.VAL_TO_CLASS.values(), np.round(class_weights, 4))))
    class_weights_tensor = torch.tensor(class_weights, dtype=torch.float32)
else:
    print("Balanced, augmented sampling: unweighted loss")
    class_weights_tensor = None
```

```python
from pytorch_lightning.callbacks import LearningRateMonitor, TQDMProgressBar, EarlyStopping, ModelCheckpoint


class LrProgressBar(TQDMProgressBar):
    """Show LR in scientific notation in the progress bar."""
    def get_metrics(self, trainer, model):
        items = super().get_metrics(trainer, model)
        for key in list(items):
            if "lr" in key.lower():
                items[key] = f"{float(items[key]):.2e}"
        return items


EPOCHS = int(os.environ.get("ECG_EPOCHS", 100))  # ECG_EPOCHS=1 for a quick smoke test
log_every_n_steps = 10

early_stop_cb = EarlyStopping(
    monitor="val_loss",   # metric to watch
    patience=10,          # stop after 10 epochs with no improvement
    mode="min",           # lower is better
    verbose=True,
)

checkpoint_cb = ModelCheckpoint(
    monitor="val_loss",
    mode="min",
    save_top_k=1,
    filename=f"ecg-{ARCH}-best-{{epoch:02d}}-{{val_loss:.4f}}",  # transfer_learning.ipynb finds it by ARCH
    verbose=True,
)

trainer = pl.Trainer(
    max_epochs=EPOCHS,
    log_every_n_steps=log_every_n_steps,
    gradient_clip_val=1.0,
    accelerator="auto",  # CUDA on Linux, MPS on Apple Silicon
    callbacks=[
        early_stop_cb,
        checkpoint_cb,
        LearningRateMonitor(logging_interval="epoch"),
        LrProgressBar(),
    ],
)
```

```python
lightning_module = EcgLightningModule(num_classes=5, arch=ARCH, learning_rate=1e-3, weight_decay=5e-5,
                                      class_weights=class_weights_tensor)
trainer.fit(lightning_module, train_dl, val_dl)
```

```python
def plot_training_history(module):
    train = module.training_step_outputs
    val = module.validation_step_outputs

    train_loss = train["loss"]
    train_acc  = train["acc"]
    val_loss   = val["loss"][1:]
    val_acc    = val["acc"][1:]
    epochs = range(1, len(train_loss) + 1)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    ax1.plot(epochs, val_loss, label="Val loss", color="red")
    ax1.plot(epochs, val_acc, label="Val acc", color="blue")    
    ax1.set_title("Validation accuracy/loss per epoch")
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Loss/Accuracy")
    ax1.legend()
    ax1.grid(True)

    ax2.plot(epochs, train_loss, label="Train loss", color="red")
    ax2.plot(epochs, train_acc, label="Train acc", color="blue")

    ax2.set_title("Training accuracy/loss per epoch")
    ax2.set_xlabel("Epoch")
    ax2.set_ylabel("Accuracy")
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

plot_training_history(lightning_module)
```

```python
from sklearn.metrics import classification_report, accuracy_score, balanced_accuracy_score, cohen_kappa_score

def print_pred_truth_comparison(truth_array, pred_array):
    class_names = [ECGDataset.VAL_TO_CLASS[i] for i in sorted(ECGDataset.VAL_TO_CLASS)]

    overall_acc      = accuracy_score(truth_array, pred_array)
    balanced_acc     = balanced_accuracy_score(truth_array, pred_array)
    kappa            = cohen_kappa_score(truth_array, pred_array)
    n_correct        = (truth_array == pred_array).sum()
    n_total          = len(truth_array)
    print(f"Overall accuracy   : {overall_acc:.4f}  ({n_correct:,} / {n_total:,} correct)")
    print(f"Balanced accuracy  : {balanced_acc:.4f}  (accounts for class imbalance)")
    print(f"Cohen's kappa      : {kappa:.4f}  (1=perfect, 0=chance)")
    print()
    print("Per-class report:")
    print(classification_report(truth_array, pred_array, target_names=class_names, digits=4))

    # Per-class accuracy breakdown
    print("Per-class accuracy:")
    for i, name in enumerate(class_names):
        mask = truth_array == i
        cls_acc = (pred_array[mask] == i).mean() if mask.sum() > 0 else float("nan")
        print(f"  {name}: {cls_acc:.4f}  (n={mask.sum():,})")
```

```python
from sklearn.metrics import confusion_matrix

def plot_confusion_matrix(truth, preds, title="Confusion Matrix", normalize=True):
    class_names = [ECGDataset.VAL_TO_CLASS[i] for i in sorted(ECGDataset.VAL_TO_CLASS)]
    cm = confusion_matrix(truth, preds)
    cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)  # row-normalised

    display = cm_norm if normalize else cm.astype(float)

    fig, ax = plt.subplots(figsize=(7, 6))
    im = ax.imshow(display, interpolation="nearest", cmap="Blues",
                   vmin=0, vmax=1 if normalize else None)
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

    ax.set_xticks(range(len(class_names)))
    ax.set_yticks(range(len(class_names)))
    ax.set_xticklabels(class_names)
    ax.set_yticklabels(class_names)
    ax.set_xlabel("Predicted label")
    ax.set_ylabel("True label")
    ax.set_title(title)

    thresh = display.max() / 2
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            value = f"{cm_norm[i, j]:.2f}" if normalize else f"{cm[i, j]:,}"
            ax.text(j, i, value,
                    ha="center", va="center",
                    color="white" if display[i, j] > thresh else "black",
                    fontsize=9)

    plt.tight_layout()
    plt.show()

plot_confusion_matrix(train_truth, train_preds, title="Train Confusion Matrix", normalize=True)
```
